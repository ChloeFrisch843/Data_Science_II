## 1. Indexing and slicing strings
Positions begin at **0** and negative indexes count from the end. Slices `text[start:stop]` exclude the stop position.

```python
city = "New York"
print(city[0])    # first character
print(city[-1])   # last character
print(city[0:3])  # characters at positions 0, 1, 2
print(len(city))  # includes the space
```

## 2. Immutability and string cleanup
Strings are **immutable**: replace a string by assigning a new value. Methods such as `.strip()` and `.lower()` return new strings; they do not modify the original string.

```python
raw = "  NEW YORK  "
clean = raw.strip()  # remove whitespace from both ends
clean = clean.title()  # title-case words
print("Original:", repr(raw))  # repr shows surrounding spaces
print("Clean:", clean)

word = "data"
word = "D" + word[1:]  # create a new string rather than changing word[0]
print(word)
```

```python
raw = "  NEW YORK  "
clean = raw.lower() 
print(clean)
```

## 3. Looping through characters and using in
A string can be iterated character by character. `in` checks whether a substring exists and returns a Boolean.

```python
responses = ["Yes", "maybe", "YES"]
for response in responses:
    if "yes" in response.lower():  # normalize case before checking
        print("Affirmative:", response)

letters = 0
for char in "data":
    letters = letters + 1
print("Count:", letters)
```

## 5. Lists versus tuples
Lists use brackets `[]` and are mutable; tuples typically use parentheses `()` and are immutable. Both are ordered and use indexing.

```python
scores = [70, 80, 90]  # list: can change
coordinates = (4, 9)  # tuple: cannot change individual entries
scores[1] = 85
print(scores)
print(coordinates[0])
print(len(scores))
```

## 6. Lists with loops and methods
`.append(x)` adds to end; `.pop(index)` removes and returns an item (default: last); `.remove(value)` deletes first matching value; `.sort()` sorts the list in place and returns `None`.

```python
scores = [90, 70, 80]
scores.append(100)  # add value
scores.pop(1)      # remove item at index 1
scores.sort()      # changes scores in place
print(scores)

cleaned = []
for text in ["  boston ", "CHICAGO", " miami"]:
    cleaned.append(text.strip().title())
print(cleaned)
```
