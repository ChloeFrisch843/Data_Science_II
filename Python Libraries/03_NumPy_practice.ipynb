## Part 2 — Data Science: Weather Station Analysis
**Questions 11–22** • A provided NumPy matrix contains **90 simulated daily records**.

Run the setup below. You do not need to write or understand the data-generation formula to complete the questions.

| Column index | Measurement | Units |
|---:|---|---|
| 0 | Day | 1–90 |
| 1 | Temperature | °C |
| 2 | Humidity | % |
| 3 | Rainfall | mm |
| 4 | Wind speed | km/h |

The main `weather` matrix contains valid simulated readings. A separate array, `temperature_with_errors`, deliberately contains **three faulty or missing values** for the cleaning questions.

```python
# Provided weather data: 90 consecutive simulated days.
# Columns: Day, Temperature_C, Humidity_pct, Rainfall_mm, Wind_kmh
rng = np.random.default_rng(2026)
days = np.arange(1, 91)
temperature = np.round(
    24 + 5.5 * np.sin(np.linspace(0, 4 * np.pi, 90))
    + rng.normal(0, 2.5, 90), 1
)
humidity = np.round(np.clip(
    75 - 0.9 * (temperature - 24) + rng.normal(0, 8, 90),
    35, 98
), 1)
rainfall = np.where(
    rng.random(90) < 0.37,
    np.round(rng.gamma(shape=2.0, scale=3.0, size=90), 1),
    0.0
)
wind = np.round(np.clip(rng.normal(15, 4.5, 90), 2, 35), 1)

weather = np.column_stack((days, temperature, humidity, rainfall, wind))
weather_columns = ["Day", "Temperature_C", "Humidity_pct",
                   "Rainfall_mm", "Wind_kmh"]

# Separate copy with three faulty/missing temperature measurements.
# Use this in Questions 19-20 only.
temperature_with_errors = weather[:, 1].copy()
temperature_with_errors[7] = -99.0
temperature_with_errors[42] = 105.0
temperature_with_errors[66] = np.nan

print("Columns:", weather_columns)
print("First five rows:")
print(weather[:5])
```

### Question 19 — Identify and clean faulty measurements

In `temperature_with_errors`, treat temperatures below 0°C, above 45°C, or equal to NaN as invalid. Print their day numbers. Make a cleaned copy replacing invalid readings with `np.nan`.

**Syntax / function hints:** `np.isnan()`, `|` for OR, `np.where()`. The temperature array has the same day order as `weather`.

```python
# Q19 — Write your solution here.
bad = ((temperature_with_errors < 0)
| (temperature_with_errors > 45)
| np.isnan(temperature_with_errors))
print("Invalid day numbers:", weather[bad, 0].astype(int))
cleaned_temps = np.where(bad, np.nan, temperature_with_errors)
print("Number of missing/invalid:", np.sum(np.isnan(cleaned_temps)))
```

### Question 22 — Standardize temperature with z-scores

Convert the original clean temperature column into z-scores. Show that the mean is approximately zero and population standard deviation is approximately one.

**Syntax / function hints:** Formula: `z = (x - mean) / std`; use `np.mean()` and `np.std()`.

```python
# Q22 — Write your solution here.
temp = weather[:,1]
z = (temp-np.mean(temp))/np.std(temp)
print("First five z-scores:", np.round(z[:5],3))
print("Mean:", np.round(np.mean(z),10))
print("Std", np.round(np.std(z),10))
```
