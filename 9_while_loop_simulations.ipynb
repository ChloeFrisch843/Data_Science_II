## 1. Review: Random Numbers and Seeds
`random.randint(a, b)` chooses an integer between `a` and `b`, **including both endpoints**.

`random.seed(number)` starts the pseudorandom generator from a specified state, helping us reproduce results in the same Python environment. A seed does **not** make the outcomes more random.

```python
import random  # random is a built-in Python module for pseudorandom experiments.

random.seed(7)  # A fixed seed makes this sequence reproducible.
print(random.randint(1, 6))  # randint(1, 6) simulates one die roll.
print(random.randint(1, 6))
```

## 2. Simulate Until a Six Appears
This experiment does not have a fixed number of rolls. A `while` loop is appropriate because we stop when the event happens.

We use a maximum-attempt safety limit in code-alongs so a loop always finishes.

```python
import random
random.seed(10)

rolls = 0
die = 0  # Start with a value that is not 6.

while die != 6 and rolls < 100:
    die = random.randint(1, 6)
    rolls = rolls + 1
    print("Roll", rolls, ":", die)

print("Rolls until six (or safety limit):", rolls)
```

## 3. Stopping at a Target: Simulating a Savings Process

## 4. Add Randomness: Variable Weekly Savings
Each week we save a random amount between $5 and $20. We stop when the total reaches $100.

```python
import random
random.seed(12)

savings = 0
weeks = 0

while savings < 100 and weeks < 100:
    deposit = random.randint(5, 20)  # Random savings between $5 and $20.
    savings = savings + deposit
    weeks = weeks + 1
    print("Week", weeks, "deposit:", deposit, "total:", savings)

print("Number of weeks:", weeks)
```

## 5. Repeat an Experiment with Nested Loops
One trial is not enough to understand a random process. Use an **outer loop** for many experiments and an **inner loop** for rolls within each experiment.

Here we repeat the experiment "roll until a 6" five times.

```python
import random
random.seed(5)

experiment = 0
results = []  # Save the number of rolls from each experiment.

while experiment < 5:
    die = 0
    rolls = 0

    while die != 6 and rolls < 100:
        die = random.randint(1, 6)
        rolls = rolls + 1

    results.append(rolls)  # .append() adds one value to a list.
    print("Experiment", experiment + 1, "needed", rolls, "rolls")
    experiment = experiment + 1

print("Roll counts:", results)
print("Average rolls:", sum(results) / len(results))  # sum() adds all values; len() counts them.
```

### Think Like a Data Scientist
Does the mean from five experiments **have** to be exactly 6? No. The theoretical mean number of rolls until the first six is 6, but a small sample can differ substantially.

A simulation estimates what might happen; it does not guarantee the exact theoretical value.

## 6. From Five Experiments to Many
We can repeat the same code with more trials and print a compact summary instead of every individual roll.

```python
import random
random.seed(15)

trials = 1000
experiment = 0
all_roll_counts = []

while experiment < trials:
    die = 0
    rolls = 0

    while die != 6 and rolls < 1000:
        die = random.randint(1, 6)
        rolls = rolls + 1

    all_roll_counts.append(rolls)
    experiment = experiment + 1

print("Experiments:", len(all_roll_counts))
print("Simulated average rolls:", round(sum(all_roll_counts) / len(all_roll_counts), 2))
print("Fewest rolls:", min(all_roll_counts))  # min() finds the smallest value.
print("Most rolls:", max(all_roll_counts))   # max() finds the largest value.
```

## 7. Simulation with an `if` Statement — Two Dice
A trial consists of rolling **two dice**, then checking whether the sum is 7. Estimate the proportion of trials that produce a 7.

The theoretical probability is **6/36 = 1/6 ≈ 0.167**, because six of the 36 equally likely ordered outcomes sum to 7.

```python
import random
random.seed(30)

trials = 1000
trial = 0
successes = 0

while trial < trials:
    first_die = random.randint(1, 6)
    second_die = random.randint(1, 6)

    if first_die + second_die == 7:
        successes = successes + 1

    trial = trial + 1

estimated_probability = successes / trials
print("Experimental P(sum = 7):", round(estimated_probability, 3))
print("Theoretical P(sum = 7):", round(1 / 6, 3))
```

### Try It — Another Event
Modify the experiment below to estimate the probability that the sum of two dice is **at least 10**. What is the theoretical probability? (There are six favorable ordered outcomes out of 36.)

```python
import random
random.seed(30)

trials = 1000
trial = 0
successes = 0

while trial < trials:
    dice1 = random.randint(1, 6)
    dice2 = random.randint(1, 6)

    if dice1 + dice2 >= 10:
        successes = successes + 1

    trial = trial + 1

print("Estimated probability:", successes / trials)
print("Theoretical probability:", 6 / 36)
```

## 8. Mini Project — A Points Game Simulation
A simple points game models the kind of repeated decisions used in many projects.

Rules for this practice version:
- Start with **20 points**.
- Roll one six-sided die each turn.
- If the die is 1 or 2, lose 5 points; otherwise gain 3 points.
- Stop when points reach **0 or less**, reach **35 or more**, or after **30 turns**.

**This is a learning simulation, not a replacement for a full interactive betting game.** It prepares you to understand game loops and stopping conditions.

```python
import random
random.seed(8)

points = 20
turn = 0

while points > 0 and points < 35 and turn < 30:
    roll = random.randint(1, 6)

    if roll <= 2:
        points = points - 5
    else:
        points = points + 3

    turn = turn + 1
    print("Turn", turn, "roll:", roll, "points:", points)

print("Final points:", points)
print("Total turns:", turn)
```

### Student Challenge — Change the Rules
Create a second version of the simulation:
1. Start at 30 points.
2. Lose 8 points when the roll is 1 or 2.
3. Gain 4 points otherwise.
4. Stop at 0 points, 50 points, or 40 turns.
5. Print the ending condition: **lost all points**, **reached the goal**, or **turn limit**.

```python
import random
random.seed(8)

points = 30
turn = 0

while points > 0 and points < 50 and turn < 40:
    roll = random.randint(1, 6)

    if roll <= 2:
        points = points - 8
    else:
        points = points + 4

    turn = turn + 1
    print("Turn", turn, "roll:", roll, "points:", points)

if points <= 0:
    print("Lost all points")
elif points >= 50:
    print("Reached the goal")
else:
    print("Turn limit")

print("Final points:", points)
print("Total turns:", turn)
```
