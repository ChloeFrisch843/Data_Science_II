```python
import random
from time import sleep

# Print important messages slowly.

def print_slow(text):
    """A function that prints one character at a time"""
    for x in text: #Cycle through the text
        print(x, end="", flush = True) #Print character, no new line
        sleep(0.02) #Delay between characters
    print() # New line
    sleep(0.5) # Delay after finishing the line

# Provide the user instructions.
def show_rules():
    print_slow("Welcome to Vegas!")
    print_slow("You have reached Lucky Dice Arcade!")
    print_slow("You start with 100 points.")
    print_slow("Roll a 1 or 2: Lose your whole bet.")
    print_slow("Roll a 3 or 4: Lose half your bet.")
    print_slow("Roll a 5: Win half your bet.")
    print_slow("Roll a 6: Win your whole bet!")
    print_slow("Type quit anytime to cash out.")


print("""
+----------------------+
|   LUCKY DICE ARCADE  |
|                      |
|     +---------+      |
|     | o     o |      |
|     |    o    |      |
|     | o     o |      |
|     +---------+      |
+----------------------+
""")

show_rules()



points = 100

# The actual game loop.

while points >= 1:
    print("You have {} points.".format(points))
    bet = input("How many points do you want to bet? (or type quit): ")
  
# How to end the game. 
    if bet == "quit":
        break
# Check for invalid input. 
    if not bet.isdecimal():
        print("Please enter a valid number.")
        continue

    bet = int(bet)

    if bet <= 0 or bet > points:
        print("Invalid bet! Try again.")
        continue
    # Roll the dice and display random message. 
    
    roll = random.randint(1,6)
    print("You rolled a", roll)
    messages = ["Good luck!", "Big reveal!:)", "LET'S GO!"]
    print(random.choice(messages))

    # Update points based on the dice roll. 
   
    if roll == 1 or roll == 2:
        points = points - bet
        print("You lost your whole bet!")

    elif roll == 3 or roll == 4:
        points = points - bet / 2
        print("You lost half your bet!")

    elif roll == 5:
        points = points + bet / 2
        print("You won half your bet!")

    else:
        points = points + bet
        print("Jackpot! You won your bet!")

# Display game over or final score screen
if points < 1:
    print_slow("GAME OVER!")
    print("You lost all your points!")
else:
    print("Thanks for playing!")
    print("Your final score is", points)


       

  



```
