## 1. Review: `for` versus `while`

- Use `for` when you want to process items in a collection or repeat a known number of times.
- Use `while` when you want to keep going **as long as a condition is true**.
- Remember to update a value used in the condition, or the loop might never stop.

```python
# Review: for loops repeat for each number in range(3).
for trial in range(3):  # range(3) gives 0, 1, 2.
    print("For loop trial:", trial)

# A while loop repeats while its condition is True.
trial = 0  # Start the counter at zero.
while trial < 3:  # Check the condition before every repetition.
    print("While loop trial:", trial)
    trial = trial + 1  # Update the counter so the loop can stop.
```

## 2. Stop When a Target Is Reached
Imagine collecting samples until the total is at least 20. We don't need to know the exact number of iterations in advance.

```python
samples_collected = 0
samples_per_trip = 6
trips = 0

while samples_collected < 20:
    samples_collected = samples_collected + samples_per_trip
    trips = trips + 1
    print("Trip", trips, "Total samples:", samples_collected)

print("Trips needed:", trips)
```

## 3. `break`: Stop Early
Sometimes the data include a **sentinel** value that tells us to stop. Here, `-1` means the sensor stopped collecting readings. The sentinel is not a temperature.

```python
readings = [21, 23, 22, -1, 25, 27]
position = 0

while position < len(readings):  # len() counts the number of items in a list.
    reading = readings[position]  # Indexing selects one reading.
    if reading == -1:
        break  # break exits the nearest loop immediately.
    print("Valid reading before stop:", reading)
    position = position + 1

print("Collection ended")
```

## 4. `continue`: Skip One Iteration
Use `continue` when you want to skip a bad observation but keep processing later observations.

**Important:** In a `while` loop, update your counter *before* reaching `continue`, or you could get stuck at the same value.

```python
measurements = [19, -5, 22, 24, -2, 21]
index = 0
valid_total = 0
valid_count = 0

while index < len(measurements):
    measurement = measurements[index]
    index = index + 1  # Move forward BEFORE a possible continue.

    if measurement < 0:
        continue  # Skip the invalid reading; continue with the next reading.

    valid_total = valid_total + measurement
    valid_count = valid_count + 1

print("Valid values:", valid_count)
print("Mean of valid values:", valid_total / valid_count)
```

### Try It — Filter Test Scores
Process `scores` using `while` and `continue`. Ignore scores less than 0 or greater than 100. Count how many valid scores remain.

```python
scores = [85, -4, 91, 103, 78, 100]
index = 0
valid_count = 0

while index < len(scores):
    score = scores[index]
    index = index + 1

    if score < 0 or score > 100:
        continue

    valid_count = valid_count + 1

print(valid_count)
```

## 5. Nested Loops: Groups of Data
A **nested loop** is a loop inside another loop. The outer loop controls the groups; the inner loop processes values within each group.

For example, these are three days of temperature readings.

```python
daily_temperatures = [
    [70, 72, 74],  # Day 1
    [68, 71, 73],  # Day 2
    [75, 77, 76]   # Day 3
]

day = 0
while day < len(daily_temperatures):
    print("Day", day + 1)
    reading_index = 0  # Reset the inner counter for every new day.

    while reading_index < len(daily_temperatures[day]):
        print("  Reading:", daily_temperatures[day][reading_index])
        reading_index = reading_index + 1

    day = day + 1  # Update the outer counter after the inner loop ends.
```

## 6. Nested Loops and Data Summary
Calculate a mean **for each day**, using a total and count that reset inside the outer loop.

```python
daily_temperatures = [[70, 72, 74], [68, 71, 73], [75, 77, 76]]
day = 0

while day < len(daily_temperatures):
    total = 0  # Start a new total for this day.
    reading_index = 0

    while reading_index < len(daily_temperatures[day]):
        total = total + daily_temperatures[day][reading_index]
        reading_index = reading_index + 1

    average = total / len(daily_temperatures[day])
    print("Day", day + 1, "average:", round(average, 1))  # round() controls displayed decimals.
    day = day + 1
```

### Try It — Weekly Sales
Use a nested loop to print the sum of sales for each store. Each inner list represents one store's three days of sales.

```python
store_sales = [[100, 120, 110], [95, 105, 125], [150, 130, 140]]
store = 0

while store < len(store_sales):
    total = 0
    day = 0

    while day < len(store_sales[store]):
        total = total + store_sales[store][day]
        day = day + 1

    print("Store", store + 1, "total sales:", total)
    store = store + 1
```

## 7. Compare Nested `for` and Nested `while`
Both loops can analyze the same grouped data. A `for` loop is often simpler when we already have every value.

```python
survey_groups = [[4, 5, 3], [3, 4], [5, 5, 4]]

for group in survey_groups:  # Each group is a list of survey ratings.
    total = 0
    for rating in group:
        total = total + rating
    print("Group average:", round(total / len(group), 2))
```

## 8. Data Science Challenge — Process Sensor Data
A sensor sends measurements in groups. `-1` means the sensor failed for that observation. An empty group would need special handling, but all groups below contain at least one valid measurement.

**Your task:** With nested `while` loops, ignore `-1`, calculate the average of valid readings for each day, and print a warning if any valid reading is greater than 30.

**Hint:** Keep the inner index moving on *every* iteration. Use `continue` after updating it.

```python
sensor_days = [[22, 24, -1, 25], [27, 31, -1, 29], [20, 21, 22]]

day = 0

while day < len(sensor_days):
    total = 0
    count = 0
    index = 0
    warning = False

    while index < len(sensor_days[day]):
        reading = sensor_days[day][index]
        index = index + 1

        if reading == -1:
            continue

        total = total + reading
        count = count + 1

        if reading > 30:
            warning = True

    average = total / count
    print("Day", day + 1, "average:", round(average, 1))

    if warning:
        print("Warning: Reading greater than 30")

    day = day + 1
```
