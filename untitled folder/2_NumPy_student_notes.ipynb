## 20. Putting It Together: Clean Scores

*Presentation slide 21*

**My Notes — copy the key ideas from the slide**

- **Which values are invalid in this example?**  
  
- **Why is `dtype=float` useful here?**  
  
- **Important syntax / details:**  
  


```python
# Topic 20: Copy the complete code example from Slide 21 below.
raw=np.array(
    [88,105,74,-6,82,100,91],
    dtype=float
)

clean = np.where(
    (raw >=0) & (raw<=100),
    raw,
    np.nan
)

print(clean)
print("Valid average:", np.nanmean(clean))
```
