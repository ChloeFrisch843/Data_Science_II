## 2. Line Plot — plt.plot()

*Presentation slide 3*

```python
# Topic 2: Copy the complete Python example from Slide 3 below.
days = [1,2,3,4,5]
temps = [70,74,72,78,81]

plt.plot(days,temps)
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Temperature Over Time")
plt.show

```

## 3. Line Plot Extras

*Presentation slide 4*

```python
# Topic 3: Copy the complete Python example from Slide 4 below.
days = [1,2,3,4,5]
week_1=[70,74,72,78,81]
week_2= [68,71,75,76,80]

plt.plot(days, week_1, marker = "8",
    linewidth=2, label = "Week 1")
plt.plot(days, week_2, marker = "8",
        linestyle ="--", label = "Week 2")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Two Weeks of Temperature")
plt.grid(True)
plt.legend()
plt.show()
         


```

## 4. Scatter Plot — plt.scatter()

*Presentation slide 5*

```python
# Topic 4: Copy the complete Python example from Slide 5 below.
study_hours = [1,2,3,4,5,6]
exam_scores = [62, 68, 74, 80, 87,91]

plt.scatter(study_hours, exam_scores,
            s =70, alpha = 0.75)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()

```

## 5. Bar Plot — plt.bar()

*Presentation slide 6*

```python
# Topic 5: Copy the complete Python example from Slide 6 below.
categories = ["A", "B", "C", "D"]
counts = [18, 25, 14, 20]

plt.bar(categories, counts)
plt.xlabel("Category")
plt.ylabel("counts")
plt.title("Survey Responses by Category")
plt.show

```

## 6. Horizontal Bar — plt.barh()

*Presentation slide 7*

```python
# Topic 6: Copy the complete Python example from Slide 7 below.
cities = ["Boston", "Chicago", "Miami", "Seattle"]
sales = [120,160,140,175]

plt.barh(cities,sales)
plt.xlabel("Sales")
plt.ylabel("City")
plt.title("Sales by City")
plt.show()

```
