## 3. Line Plot Extras

*Presentation slide 4*

```python
# Topic 3: Copy the complete Python example from Slide 4 below.
days = [1,2,3,4,5]
week_1=[70,74,72,78,81]
week_2= [68,71,75,76,80]

plt.plot(days, week_1, marker = "8",
    linewidth=2, label = "Week 1")
plt.plot(days, week_2, marker = "8",
        linestyle ="--", label = "Week 2")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Two Weeks of Temperature")
plt.grid(True)
plt.legend()
plt.show()
         


```

```python
# Topic 4: Copy the complete Python example from Slide 5 below.
study_hours = [1,2,3,4,5,6]
exam_scores = [62, 68, 74, 80, 87,91]

plt.scatter(study_hours, exam_scores,
            s =70, alpha = 0.75)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()

```

## 10. Figure Size + Axis Controls

```python
# Topic 10: Copy the complete Python example from Slide 11 below.
months = [1, 2, 3, 4, 5, 6]
revenue = [20, 24, 29, 27, 34, 39]
plt.figure(figsize=(8, 4))
plt.plot (months, revenue, marker="o")
plt.xlim (1, 6)
plt.ylim (15, 45)
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Revenue by Month")
plt. show()
```

## 11. Reference Lines

*Presentation slide 12*

```python
# Topic 11: Copy the complete Python example from Slide 12 below.
months = [1, 2, 3, 4, 5, 6]
revenue = [20, 24, 29, 27, 34, 39]
plt.plot(months, revenue, marker="o") 
plt. axhline(y=30, linestyle="--",
label="Goal = 30")
plt.axvline(x = 4, color= "gray", linestyle = ":", 
            label= "Month 4")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Revenue with Reference Lines")
plt. legend()
plt.grid(True)
plt.show()

```

## 13. Multiple Plots — plt.subplots()

*Presentation slide 14*

```python
# Topic 13: Copy the complete Python example from Slide 14 below.
days = [1, 2, 3, 4, 5]
temps = [70, 74, 72, 78, 81]
study_hours = [1, 2, 3, 4, 5, 6]
exam_scores = [62, 68, 74, 80, 87, 91]
fig, axes = plt.subplots(1, 2, figsize= (9, 4))
axes [0]. plot (days, temps, marker="o") 
axes [0]. set_title("Temperature")
axes [0]. set_xlabel("Day")
axes [0]. set_ylabel("Temperature")
axes [1]. scatter (study_hours, exam_scores) 
axes [1]. set_title("Study vs Score") 
axes [1]. set_xlabel("Study Hours") 
axes [1]. set_ylabel("Exam Score")
fig. tight_layout()
plt. show()

```
