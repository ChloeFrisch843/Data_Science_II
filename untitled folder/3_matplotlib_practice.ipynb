## 1. Import Matplotlib & Create the Dataset
The cell below uses Pandas and NumPy to construct 90 days of café activity. **Run it as provided; you do not need to understand every line of the data-generation code.**

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------- DATASET SETUP: RUN THIS CELL ----------------
rng = np.random.default_rng(42)
n = 90

day = np.arange(1, n + 1)
dates = pd.date_range("2026-02-01", periods=n, freq="D")
weekday = dates.day_name()
weekend = np.isin(dates.dayofweek, [5, 6])

weather = rng.choice(["Sunny", "Cloudy", "Rainy"], size=n,
                     p=[0.47, 0.34, 0.19])
promotion = rng.choice(["None", "Student Discount", "BOGO", "Social Media"],
                       size=n, p=[0.42, 0.26, 0.18, 0.14])

sun_bonus = np.where(weather == "Sunny", 4,
                     np.where(weather == "Rainy", -5, 0))
temp = np.clip(43 + 0.31 * day + sun_bonus + rng.normal(0, 6, n), 30, 87)

promo_bonus = np.where(promotion == "None", 0, 14)
customers = np.clip(np.rint(98 + 22 * weekend + promo_bonus +
                           0.8 * (temp - 55) + sun_bonus +
                           rng.normal(0, 13, n)), 45, 190).astype(int)

hot_drinks = np.clip(np.rint(0.47 * customers - 0.95 * (temp - 55)
                              + rng.normal(0, 7, n)), 5, None).astype(int)
iced_drinks = np.clip(np.rint(0.38 * customers + 1.3 * (temp - 55)
                               + rng.normal(0, 7, n)), 5, None).astype(int)
pastries = np.clip(np.rint(0.44 * customers + rng.normal(0, 8, n)),
                   5, None).astype(int)

ad_spend = np.clip(np.rint(24 + np.where(promotion == "None", 0, 28) +
                             rng.normal(0, 9, n)), 5, None).astype(int)
wait = np.clip(2.7 + 0.047 * (customers - 85) +
               0.4 * (promotion != "None") + rng.normal(0, 0.55, n), 1.5, 12)
satisfaction = np.clip(4.75 - 0.17 * (wait - 3) +
                       rng.normal(0, 0.24, n), 2.0, 5.0)

revenue = np.rint(hot_drinks * 4.50 + iced_drinks * 5.25 +
                  pastries * 4.10 + customers * 2.4 +
                  rng.normal(0, 33, n)).astype(int)
expenses = np.rint(0.56 * revenue + 165 + 0.70 * ad_spend +
                   rng.normal(0, 19, n)).astype(int)
profit = revenue - expenses
social_views = np.clip(np.rint(rng.lognormal(7.0, 1.24, n) *
                                np.where(promotion == "Social Media", 2.5, 1)),
                       30, 60000).astype(int)

df = pd.DataFrame({
    "Day": day,
    "Date": dates,
    "Weekday": weekday,
    "Weekend": np.where(weekend, "Yes", "No"),
    "Weather": weather,
    "Promotion": promotion,
    "Temperature_F": np.round(temp, 1),
    "Customers": customers,
    "Hot_Drinks": hot_drinks,
    "Iced_Drinks": iced_drinks,
    "Pastries": pastries,
    "Avg_Wait_Min": np.round(wait, 1),
    "Satisfaction_1to5": np.round(satisfaction, 2),
    "Ad_Spend_USD": ad_spend,
    "Revenue_USD": revenue,
    "Expenses_USD": expenses,
    "Profit_USD": profit,
    "Social_Views": social_views,
})

print(f"Dataset loaded: {len(df)} rows and {len(df.columns)} columns")
df.head(6)
````

### Dataset dictionary
Each row represents **one day** at the café. Values such as revenue, temperature, and customer counts are generated to show plausible patterns, but are not actual business results.

| Column | What it measures | Column | What it measures |
|---|---|---|---|
| `Day` | Day number (1–90) | `Date` | Calendar date |
| `Weekday` | Day of the week | `Weekend` | Yes or No |
| `Weather` | Sunny, Cloudy, or Rainy | `Promotion` | Type of promotion that day |
| `Temperature_F` | Temperature in °F | `Customers` | Number of customers |
| `Hot_Drinks` | Hot drinks sold | `Iced_Drinks` | Iced drinks sold |
| `Pastries` | Pastries sold | `Avg_Wait_Min` | Average wait time, minutes |
| `Satisfaction_1to5` | Average satisfaction, 1–5 | `Ad_Spend_USD` | Daily ad spending, USD |
| `Revenue_USD` | Daily revenue, USD | `Expenses_USD` | Daily expenses, USD |
| `Profit_USD` | Revenue minus expenses, USD | `Social_Views` | Daily social-media views |

## 16. Matplotlib Cheat Sheet — Quick Reference

| Plot / technique | Function |
|---|---|
| Line · Scatter · Vertical bar · Horizontal bar | `plt.plot()` · `plt.scatter()` · `plt.bar()` · `plt.barh()` |
| Histogram · Box plot · Pie chart | `plt.hist()` · `plt.boxplot()` · `plt.pie()` |
| Labels · Title · Legend · Grid | `plt.xlabel()` · `plt.ylabel()` · `plt.title()` · `plt.legend()` · `plt.grid()` |
| Figure size · Ranges · Tick labels | `plt.figure(figsize=(...))` · `plt.xlim()` · `plt.ylim()` · `plt.xticks()` |
| Horizontal / vertical reference line | `plt.axhline()` · `plt.axvline()` |
| Multiple plots · Save | `plt.subplots()` · `fig.tight_layout()` · `plt.savefig()` |

**Reminder:** A value like `...` in the slides' cheat sheet is a placeholder, not an actual histogram bin count. Use a value such as `bins=8`.
